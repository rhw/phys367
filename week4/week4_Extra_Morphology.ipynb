## "Sandbox" notebook for morphology / size studies

#### Standard imports

```python
import os
import tables_io
import numpy as np
import matplotlib.pyplot as plt
import pyccl as ccl
```

#### Change this to point at the correct locations

```python
pz_dir = '.'
```

#### Read some training data

```python
t = tables_io.read(f'{pz_dir}/data/lsst_cat_matched_nonan_train.hdf5')
```

#### Extract some size information

```python
def r_90_of_n(n):
    # Approximate ratio r_90 / r_50 for a Sersic profile with index n.
    # Multiplying the half-light radius by this factor gives the 90%-light radius.
    return 1.25 + 1.125*n
```

```python
sersic_x = t['sersic_reff_x']
sersic_y = t['sersic_reff_y']
sersic_index = t['sersic_index']           # n=1: exponential disk, n=4: de Vaucouleurs elliptical
sersic_ratio = r_90_of_n(sersic_index)    # r_90 / r_50 factor
# Sersic chord ≈ r_50 in arcsec (geometric mean of the two semi-axes)
sersic_chord = np.sqrt(sersic_x*sersic_x + sersic_y*sersic_y)
spiral_mask = sersic_index < 3            # disk-dominated galaxies
ellip_mask = sersic_index > 3             # bulge-dominated / elliptical galaxies
```

#### Make a standard cosmology and use it to compute some useful quantities

```python
# Standard flat ΛCDM cosmology. BBKS transfer function is a fast approximation
# adequate for computing angular diameter distances over the redshift range here.
cosmo = ccl.Cosmology(
    Omega_c=0.25, Omega_b=0.05,
    h=0.7, n_s=0.95, sigma8=0.8,
    transfer_function='bbks'
)
z_grid = np.linspace(0., 6., 301)[1:]
a_grid = (1/(1+z_grid))                           # scale factor
D_a = cosmo.angular_diameter_distance(a_grid)     # angular diameter distance [Mpc]
rad_to_asec = 360*60*60/(2*np.pi)                 # radians → arcseconds
ang_scale_z = D_a * 1000/ rad_to_asec            # kpc per arcsecond vs redshift
dist_mod = cosmo.distance_modulus(a_grid)          # distance modulus μ(z)
```

#### Here is a standard quality mask

```python
mask  = (t['refExtendedness'] == 1) *\
        (t['g_psfFlux']/ t['g_psfFluxErr'] > 10) *\
        (t['i_psfFlux'] / t['i_psfFluxErr'] > 10) *\
        (t['r_psfFlux'] / t['r_psfFluxErr'] > 10) *\
        (t['z_psfFlux'] / t['z_psfFluxErr'] > 10)

```

#### The next few cells extract a variety of useful quantities

```python
mag = t['i_sersicMag']
psfmag = t['i_psfMag']
```

```python
redshifts = t['redshift']
# Clamp redshifts to a valid range; 9 is a sentinel for missing/bad spec-z.
redshifts = np.where(np.isfinite(redshifts), redshifts, 9).clip(0.0001, 10.)

scale = 1/(1+redshifts)    # scale factor per object
scale = np.where(np.isfinite(scale), scale, 0.1)
```

```python
D_a_vals = cosmo.angular_diameter_distance(scale)
rad_to_asec = 360*60*60/(2*np.pi) 
ang_scale = D_a_vals * 1000/ rad_to_asec
dist_mods = cosmo.distance_modulus(scale)
```

```python
# Second moment trace T = Ixx + Iyy per band.
# For a circular Gaussian PSF, sqrt(T)*pixel_scale gives the RMS radius in arcsec.
# Comparing T across bands reveals wavelength-dependent morphology (larger in bluer bands
# for star-forming galaxies due to extended blue knots).
traces = {}
for band in 'ugrizy':
   traces[band] = t[f"{band}_ixx"] + t[f"{band}_iyy"]
```

```python
chords = {}
for band in 'ugrizy':
   chords[band] = np.sqrt(t[f"{band}_ixx"] + t[f"{band}_iyy"])

chords2d = np.vstack( [chords[band] for band in 'griz'] )
chord_scatter = np.var(chords2d, axis=0)
chord_mean = np.mean(chords2d, axis=0)
```

```python
# g/z size ratio: log10(chord_g / chord_z). Positive → larger in g (bluer = more extended SF).
# Negative → smaller in g (passive/elliptical with central red bulge dominating).
g_z_size_ratio = np.log10(chords['g'] / chords['z'])

# Color difference between PSF mag (unresolved core) and fixed aperture mag:
# large positive value means the core is bluer than the outer galaxy (or vice versa).
g_z_1p0_color_diff = (t['g_psfMag'] - t['z_psfMag']) - (t['g_gaap1p0Mag'] - t['z_gaap1p0Mag'])
g_z_1p5_color_diff = (t['g_psfMag'] - t['z_psfMag']) - (t['g_gaap1p5Mag'] - t['z_gaap1p5Mag'])
g_z_2p5_color_diff = (t['g_psfMag'] - t['z_psfMag']) - (t['g_gaap2p5Mag'] - t['z_gaap2p5Mag'])
g_z_3p0_color_diff = (t['g_psfMag'] - t['z_psfMag']) - (t['g_gaap3p0Mag'] - t['z_gaap3p0Mag'])
size_mask = np.bitwise_and(
    np.bitwise_and(
        np.isfinite(g_z_size_ratio[mask]),
        np.isfinite(g_z_2p5_color_diff[mask]),
    ),
    np.isfinite(g_z_3p0_color_diff[mask]),
)
```

#### Now we make a bunch of scatter plots, looking for correlations

```python
correl = np.corrcoef(g_z_size_ratio[mask][size_mask], g_z_1p0_color_diff[mask][size_mask])[1,0]
_ = plt.scatter(g_z_size_ratio[mask], g_z_1p0_color_diff[mask], c=redshifts[mask], label=f"Correl = {correl:0.3f}")
_ = plt.xlim(-1, 1)
_ = plt.ylim(-1, 1)
_ = plt.legend()
```

```python
#_ = plt.scatter((t['g_psfMag'] - t['z_psfMag'])[mask], g_z_1p5_color_diff[mask], c=redshifts[mask], s=1)
_ = plt.hist2d((t['g_psfMag'] - t['z_psfMag'])[mask], g_z_1p5_color_diff[mask], bins=(np.linspace(-1,4, 101), np.linspace(-1,1,101)), norm='log')

_ = plt.xlim(-1, 4)
_ = plt.ylim(-1, 1)
_ = plt.colorbar()
```
